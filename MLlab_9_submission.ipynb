```python
import warnings
warnings.filterwarnings("ignore")
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans , AgglomerativeClustering
from sklearn.decomposition import PCA
from sklearn.metrics import (silhouette_score, davies_bouldin_score,calinski_harabasz_score)

from scipy.cluster.hierarchy import linkage , dendrogram
sns.set_theme(style = "whitegrid")
plt.rcParams['figure.figsize'] = (8,5)


df = pd.read_csv("/content/Wholesale customers data.csv")
df.head()
```

```python
features = ["Fresh","Milk","Grocery","Frozen","Detergents_Paper","Delicassen"]
X = df[features].copy()
display(X.head())
print("Number of records used for clustering:",len(X))
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
X_scaled_df = pd.DataFrame(X_scaled , columns=features , index = X.index)
```

```python
print("First five scaled records:")
display(X_scaled_df.head())
```

```python
#Calculate WCSS (Within cluster sum of squares)
wcss = []
k_values = range(1,11)
for k in k_values:
  kmeans = KMeans(n_clusters=k , random_state=42, n_init=10)
  kmeans.fit(X_scaled)
  wcss.append(kmeans.inertia_)

#Plot the Elbow curve
plt.figure(figsize=(8,5))
plt.plot(list(k_values),wcss,marker="o")
plt.title("Elbow Method for optimal number of clusters")
plt.xlabel("Number of clusters(K)")
plt.ylabel("WCSS/Interia")
plt.xticks(list(k_values))
plt.grid(True)
plt.show()
```

```python
silhouette_values = []
for k in range(2,11):
  model = KMeans(n_clusters=k , random_state=42, n_init=10)
  labels= model.fit_predict(X_scaled)
  score = silhouette_score(X_scaled, labels)
  silhouette_values.append(score)

silhouette_table = pd.DataFrame({
    "Number of clusters":list(range(2,11)),
    "Silhouette_score":silhouette_values})
display(silhouette_table.round(4))
```

```python
suggested_k = int(
    silhouette_table.loc[silhouette_table["Silhouette_score"].idxmax(),"Number of clusters"]
)
print("K suggested by the highest silhouette score:",suggested_k)
```

```python
K = suggested_k

kmeans_model = KMeans(n_clusters=K , random_state=42, n_init=10)
kmeans_labels = kmeans_model.fit_predict(X_scaled)
```

```python
df_kmeans = df.copy()
df_kmeans["KMeans_Cluster"] = kmeans_labels

print("K-means clustering completed")
print("\nNumber of customers in each cluster:")
print(df_kmeans["KMeans_Cluster"].value_counts().sort_index())

print("\nSample records with cluster labels:")
display(df_kmeans.head())
```

```python
hierarchical_model = AgglomerativeClustering(
    n_clusters=K,
    linkage="ward",
    metric="euclidean"
)
hierarchical_labels = hierarchical_model.fit_predict(X_scaled)
```

```python
df_hierarchical = df.copy()
df_hierarchical["Hierarchical_Cluster"] = hierarchical_labels

print("Hierarchical clustering completed")
print("\nNumber of customers in each cluster:")
print(
    df_hierarchical["Hierarchical_Cluster"].value_counts().sort_index()
)

print("\nSample records with cluster labels:")
display(df_hierarchical.head())
```

```python
Z = linkage(X_scaled,method="ward",metric="euclidean")
plt.figure(figsize=(16,7))
dendrogram(Z,leaf_rotation=90,leaf_font_size=6)
plt.title("Dendrogram - Hierarchical Clustering")
plt.xlabel("Customer Index")
plt.ylabel("Ward Linkage Distance")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12,6))
dendrogram(Z,truncate_mode="lastp",p=30,show_leaf_counts=True,leaf_rotation=45)
plt.title("Truncated Dendrogram - last30 Merged Groups")
plt.xlabel("Cluster or Group")
plt.ylabel("Ward Linkage Distance")
plt.tight_layout()
plt.show()
```

```python
#reduce six scaled features to two principal components
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

print(
    "Variance explained by the two PCA components:",
    round(pca.explained_variance_ratio_.sum() * 100,2),
    "%"
)

#create a plotting dataframe
plot_df = pd.DataFrame({"PC1":X_pca[:,0],"PC2":X_pca[:,1],"KMeans_Cluster":kmeans_labels.astype(str),
                "Hierarchical_Cluster":hierarchical_labels.astype(str)
          })
```

```python
#k-means visualization
plt.figure(figsize=(9,6))
sns.scatterplot(data=plot_df,x="PC1",y="PC2",hue="KMeans_Cluster",palette="tab10",s=70)
plt.title("Customer Segments - K-means Clustering")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.legend(title="Cluster")
plt.tight_layout()
plt.show()
```
